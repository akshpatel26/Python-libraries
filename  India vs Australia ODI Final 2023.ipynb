```python
plt.figure(figsize=(12,6))
plt.bar(india_batting_df.index,india_batting_df['Runs'],label='India',color='lightblue')
for i, (runs, balls) in enumerate(zip(india_batting_df['Runs'], india_batting_df['Balls'])):
    plt.text(i, runs, f'{runs} ({balls})', ha='center', va='bottom', color='black')
plt.title("Runs Scored: India")   
plt.xlabel("Batsmen",color="r")
plt.ylabel("Runs",color="r")
plt.xticks(range(len(india_batting_df)), india_batting_df['Batsman'], rotation=60)
plt.legend()
    
plt.show()
```

```python

plt.figure(figsize=(10, 6))

# Create the barplot using seaborn
sns.barplot(x="Econ", y="Bowler", data=aus_bowling_df, palette="viridis")

for index, row in aus_bowling_df.iterrows():
    plt.text(row["Econ"] + 0.05, index, f'{row["Econ"]:.2f}', color="black", ha="left", va="center", fontsize=10)
plt.title("Australia Bowling: Economy Rate")
plt.xlabel("Economy Rate")
plt.ylabel("Bowlers")
plt.grid(True)
plt.show()
```

```python
# India Bowling: Economy Rate
plt.figure(figsize=(10,6))
bars = plt.barh(india_bowling_df["Bowler"],india_bowling_df["Econ"],color="skyblue")

for bar in bars:
    width = bar.get_width()
    plt.text(width,bar.get_y() + bar.get_height()/2,'{:.2f}'.format(width),ha= "left",va="center",color="black",fontsize=10)

plt.title("India Bowling: Economy Rate")
plt.xlabel("Economy Rate")
plt.ylabel("Bowlers")
plt.grid(True)
plt.show()
```

```python
dismissal_counts = india_batting_df['Dismissal'].value_counts()

# Display dismissal trend analysis
print("\nDismissal Trends:")
print(dismissal_counts)

#group dismissals
dismissal_type = india_batting_df['Dismissal'].apply(lambda x: x.split()[0])
dismissal_type_counts = dismissal_type.value_counts()

print("\nDismissal Type Counts:")
print(dismissal_type_counts)

plt.figure(figsize=(10, 6))
dismissal_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Dismissal Trends in India Batting')
plt.xlabel('Dismissal Type')
plt.ylabel('Frequency')
plt.xticks(rotation=45, ha='right')
plt.show()
```
